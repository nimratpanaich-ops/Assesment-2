1.1 Problem 1: Tip Calculator
1.1.1 The Scenario
You work at a Perth restaurant. The manager wants a simple tool where staff can enter:

The bill amount (e.g., \$45.80)
The tip percentage (e.g., 15%)
And get back:

The tip amount
The total amount to charge
1.1.2 What You Must Submit
STEP 1: Understand the Problem

In your own words, what is this program supposed to do? (2-3 sentences)

[This program is made to help restaurant staff quickly calculate tips for customers. The staff member enters the bill amount and the tip percentage, and the program works out how much the tip is and what the final total amount will be after adding the tip. This helps avoid mistakes and saves time during busy restaurant hour, mainly on weekends.]

STEP 2: Identify Inputs and Outputs

INPUTS:.Bill amount entered by the staff member
.Tip percentage chosen by the customer
.Both inputs are numbers. The bill amount can include decimals because restaurant bills usually have cents (for example: \$45.80). The tip percentage is normally a whole number like 10%, 15%, or 20%.

[What does the user provide?]
[What type of data? Numbers? Text?]
OUTPUTS:.The tip amount in dollars
.The final total amount including the tip
.The amounts should be shown in dollar format (for example: \$6.87,\$2.5)

[What does the program show back?]
[What format?]
STEP 3: Work It By Hand

Show me you can do this with a calculator. Use this example:

Bill amount: \$45.80
Tip percentage: 15%
Write the steps:.First, convert 15% into decimal form:15% = 0.15
.Multiply the bill amount by the tip percentage:\$45.80 × 0.15 = \$6.87
.Add the tip amount to the original bill:\$45.80 + \$6.87 = \$52.67
.Final answer I would tell the staff member:“The tip amount is \$6.87 and the total bill including tip is \$52.67.”
