# Module 1 – Verification Through Testing
In this module, I learned that verification is important because it helps check whether the code is working correctly. The module explained that programmers should not only write code but also test and verify the results properly. 

I learned about assertions, which are simple checks used to compare expected and actual outputs. If the result matches, the test passes, and if it does not, it shows there is an error in the program. 

The module also taught that AI-generated code should not be trusted without testing. Even if AI creates code quickly, it is still important to verify whether the solution is correct, simple, and understandable. 

The password checker practice challenge helped me understand how requirements can be turned into tests. It showed how verification and assertions can be used to check different scenarios and improve confidence in the program. 

Overall, this module improved my understanding of testing, verification, and the importance of checking code carefully before accepting the final solution.



# Module 2 – Debugging as Verification


In this module, I learned that debugging is the process of finding and understanding why a program is not working correctly. Instead of randomly fixing errors, the module explained that debugging should be done step by step by identifying the problem, checking the output, and finding the root cause. 

I also learned how important error messages are in Python. Error messages help identify what went wrong, where the error happened, and how the program reached that point. Understanding these messages makes debugging easier and improves problem-solving skills. 

Another important concept was using print statements to trace values inside the program. Print statements help check what the code is actually doing and make it easier to identify incorrect logic or unexpected outputs. 

The module also explained that AI should be used as a support tool for explanations and guidance instead of relying on it to automatically fix everything. It is important to understand the cause of the problem before accepting a solution. 

Overall, this module improved my understanding of debugging, reading error messages, and solving problems systematically. It also showed me the importance of verifying fixes carefully before considering the program complete.
