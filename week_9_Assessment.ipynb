```python
import pandas as pd

# Load dataset
sales_df = pd.read_csv('sales_data.csv')

# Display first 5 rows
print("\n📋 Sales Data Sample")
print("=" * 60)
print(sales_df.head())

# -----------------------------------
# Data Cleaning
# -----------------------------------

# Remove $ signs and commas
sales_df['Quantity'] = sales_df['Quantity'].replace('[\$,]', '', regex=True).astype(int)

sales_df['Unit_Price'] = sales_df['Unit_Price'].replace('[\$,]', '', regex=True).astype(float)

sales_df['Total_Sale'] = sales_df['Total_Sale'].replace('[\$,]', '', regex=True).astype(float)

# -----------------------------------
# Regional Sales Analysis
# -----------------------------------

regional_summary = sales_df.groupby('Region').agg(
    Total_Sales=('Total_Sale', 'sum'),
    Average_Sale=('Total_Sale', 'mean'),
    Number_of_Transactions=('Total_Sale', 'count')
).reset_index()

# Round average sales
regional_summary['Average_Sale'] = regional_summary['Average_Sale'].round(2)

# Sort by total sales
regional_summary = regional_summary.sort_values(
    by='Total_Sales',
    ascending=False
)

# Display regional summary
print("\n📊 Sales Performance by Region")
print("=" * 60)
print(regional_summary)

# -----------------------------------
# Best Performing Region
# -----------------------------------

best_region = regional_summary.loc[
    regional_summary['Total_Sales'].idxmax()
]

print("\n🏆 Best Performing Region")
print("=" * 60)

print(f"Region: {best_region['Region']}")
print(f"Total Sales: ${best_region['Total_Sales']:.2f}")
print(f"Average Sale: ${best_region['Average_Sale']:.2f}")
print(f"Transactions: {best_region['Number_of_Transactions']}")

# -----------------------------------
# Sales Representative Analysis
# -----------------------------------

sales_rep_summary = sales_df.groupby('Sales_Rep').agg(
    Total_Sales=('Total_Sale', 'sum')
).reset_index()

sales_rep_summary = sales_rep_summary.sort_values(
    by='Total_Sales',
    ascending=False
)

print("\n👨‍💼 Sales by Representative")
print("=" * 60)
print(sales_rep_summary)

# -----------------------------------
# Top Selling Products
# -----------------------------------

top_products = sales_df.sort_values(
    by='Total_Sale',
    ascending=False
)

print("\n🛒 Top Selling Products")
print("=" * 60)
print(top_products[['Product', 'Total_Sale']].head())

# -----------------------------------
# Business Insights
# -----------------------------------

print("\n📌 Business Insights")
print("=" * 60)

print(f"- The best performing region is {best_region['Region']} based on total sales.")
print("- High-value technology products generate the most revenue.")
print("- Lower-cost office products sell in larger quantities.")
print("- NSW contributes the highest company revenue.")
print("- The business should maintain stock for high-demand products.")
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv('transactions.csv')

# Clean Amount column
df['Amount'] = df['Amount'].replace('[$]', '', regex=True).astype(float)

# Calculate total spending
total_spending = df['Amount'].sum()

# Spending by category
category_spending = df.groupby('Category')['Amount'].sum()

# Calculate percentage of total spending
spending_percentage = (category_spending / total_spending) * 100

print("Spending Percentage by Category:")
print(spending_percentage.round(2))

# Identify categories above typical spending levels
print("\nPotential Savings Areas:")

for category, percentage in spending_percentage.items():
    if percentage > 20:
        print(f"- {category}: {percentage:.2f}% of total spending")
        print("  Suggestion: Consider reducing spending in this category.")

# Encouraging feedback
print("\nPositive Spending Habits:")

for category, percentage in spending_percentage.items():
    if percentage < 10:
        print(f"- Great job managing your {category} spending!")

# User-friendly financial advice
print("\nPersonal Finance Advice:")

if 'Entertainment' in spending_percentage:
    print("- Reducing entertainment expenses could improve savings.")

if 'Coffee' in spending_percentage:
    print("- Making coffee at home occasionally may reduce daily expenses.")

if 'Groceries' in spending_percentage:
    print("- Planning grocery shopping may help manage food costs.")
```
