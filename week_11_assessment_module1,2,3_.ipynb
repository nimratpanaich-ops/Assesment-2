```python
# ==========================================
# Module 1 – Practice Challenge
# Verification Through Testing
# ==========================================

# ------------------------------------------
# Challenge 1 – Addition Function
# ------------------------------------------

def add(a, b):
    return a + b


# Assertions
assert add(2, 3) == 5
assert add(-1, 1) == 0
assert add(0, 0) == 0

print("Addition function tests passed.")


# ------------------------------------------
# Challenge 2 – Discount Calculator
# ------------------------------------------

def calculate_discount(price, percent):
    return price * (percent / 100)


# Assertions
assert calculate_discount(100, 10) == 10
assert calculate_discount(50, 20) == 10
assert calculate_discount(200, 5) == 10

print("Discount calculator tests passed.")


# ------------------------------------------
# Challenge 3 – Password Strength Checker
# ------------------------------------------

def is_strong_password(password):

    # Password must be at least 8 characters
    if len(password) < 8:
        return False

    has_number = False
    has_uppercase = False

    # Check characters
    for char in password:

        if char.isdigit():
            has_number = True

        if char.isupper():
            has_uppercase = True

    return has_number and has_uppercase


# Assertions
assert is_strong_password("Pass123") == False
assert is_strong_password("password1") == False
assert is_strong_password("Password1") == True
assert is_strong_password("PASSWORD1") == True
assert is_strong_password("") == False
assert is_strong_password("Pass") == False

print("Password strength tests passed.")


# ------------------------------------------
# Challenge 4 – Temperature Converter
# ------------------------------------------

def celsius_to_fahrenheit(celsius):

    fahrenheit = (9 / 5) * celsius + 32

    return fahrenheit


# Assertions
assert celsius_to_fahrenheit(0) == 32
assert celsius_to_fahrenheit(100) == 212
assert celsius_to_fahrenheit(25) == 77
assert celsius_to_fahrenheit(-40) == -40

print("Temperature conversion tests passed.")


# ------------------------------------------
# Final Output
# ------------------------------------------

print("\nAll Module 1 practice challenge tests passed successfully.")
```

```python
# ==========================================
# Debugging Challenge – Vowel Counter
# ==========================================

def count_vowels(word):

    vowels = "aeiou"
    count = 0

    for letter in word:

        if letter in vowels:
            count += 1

    return count


# ------------------------------------------
# Verification Tests
# ------------------------------------------

assert count_vowels("Audio") == 3
assert count_vowels("hello") == 2
assert count_vowels("python") == 1
assert count_vowels("") == 0


print("All vowel counter tests passed.")


# ------------------------------------------
# Example Output
# ------------------------------------------

print(count_vowels("Audio"))       # 3
print(count_vowels("education"))   # 5
```

```python
# ==========================================
# Module 3 – Strategic Test Design
# ==========================================

# ------------------------------------------
# Challenge – Phone Number Validator
# ------------------------------------------

def validate_phone_number(phone):

    # Remove spaces, dashes, and brackets
    cleaned = ""

    for char in phone:

        if char.isdigit():
            cleaned += char

    # Check length
    if len(cleaned) != 10:
        return "Invalid phone number"

    # Format phone number
    formatted = "(" + cleaned[0:3] + ") " + cleaned[3:6] + "-" + cleaned[6:10]

    return formatted


# ------------------------------------------
# Verification Tests
# ------------------------------------------

assert validate_phone_number("0412345678") == "(041) 234-5678"

assert validate_phone_number("0412 345 678") == "(041) 234-5678"

assert validate_phone_number("0412-345-678") == "(041) 234-5678"

assert validate_phone_number("(041)2345678") == "(041) 234-5678"

assert validate_phone_number("12345") == "Invalid phone number"

assert validate_phone_number("abcdefghij") == "Invalid phone number"

assert validate_phone_number("") == "Invalid phone number"


print("All phone number validation tests passed.")


# ------------------------------------------
# Example Output
# ------------------------------------------

print(validate_phone_number("0412 345 678"))
print(validate_phone_number("0412-345-678"))
print(validate_phone_number("12345"))
```
