# Module 1 – Verification Through Testing

## What I Learned

In this module, I learned that testing is used to verify whether code works correctly or not. Instead of trusting AI-generated code directly, we should always test and verify the behaviour.

I also learned that assertions act like simple checks which define what “correct” means.

Example:
- A strong password should:
  - be at least 8 characters
  - contain a number
  - contain an uppercase letter

## Practice Challenge

Password: "Pass123"
Result: Weak password because it is less than 8 characters.

Password: "password1"
Result: Weak password because it does not contain an uppercase letter.

Password: "Password1"
Result: Strong password because it meets all requirements.

## Reflection

This activity helped me understand how specifications and testing work together. Before writing code, we should first define expected behaviour clearly.

# Module 2 – Debugging as Verification

## What I Learned

In this module, I learned that debugging means investigating why code is not working correctly.

Instead of randomly changing code, we should:
1. Identify the problem
2. Observe the behaviour
3. Form a hypothesis
4. Test the hypothesis
5. Fix the issue

I also learned that print statements help us understand what is happening inside the program.

## Practice Challenge

Problem:
The function for averaging positive numbers was giving incorrect results.

Expected Output:
3.0

Observed Output:
1.67

Reason:
The function was including negative numbers in the calculation.

Solution:
Filter positive numbers before calculating the average.

## Reflection

This module taught me that debugging is a step-by-step investigation process. AI can help explain errors, but we should still understand the problem ourselves.

# Module 3 – Strategic Test Design

## What I Learned

In this module, I learned that test design is about deciding what behaviour should be tested.

Good testing includes:
- happy path testing
- boundary testing
- invalid input testing
- edge cases

## Practice Challenge

Grade System:
- 90–100 = A
- 80–89 = B
- 70–79 = C
- 60–69 = D
- Below 60 = F

### Test Cases

| Input | Expected Grade |
|------|------|
| 95 | A |
| 85 | B |
| 75 | C |
| 65 | D |
| 50 | F |

### Boundary Tests

| Input | Expected Result |
|------|------|
| 90 | A |
| 89 | B |
| 60 | D |
| 59 | F |

### Invalid Inputs

- Negative numbers
- Scores above 100
- Text instead of numbers

## Reflection

This module helped me understand how to design better test cases and think about situations where systems may fail.
